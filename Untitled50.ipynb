````python
import psycopg2
from google.colab import userdata
from langchain_core.tools import tool


@tool
def evaluate_solana_unit_handover(
    tranche_id: str,
    composite_fri: float,
    fri_threshold: float,
    displacement_x: float,
    velocity_v: float,
    principal_base_units: str,
    current_base_units: str,
    best_coupled_node: str = "QQQ",
    coupled_score: float = 0.4216,
) -> str:
    """
    Evaluates Composite FRI patience gate, crest handover (x >= 0.50, v <= 0),
    exact integer base-unit principal protection, and surplus routing.
    """
    principal = int(principal_base_units)
    current = int(current_base_units)

    if principal < 0 or current < 0:
        raise ValueError("Base-unit amounts must be non-negative integers.")

    if composite_fri < fri_threshold:
        return (
            f"[{tranche_id}] YELLOW_LIGHT_HOLD -> BLOCK_HOLD | "
            f"FRI {composite_fri:.4f} < {fri_threshold:.4f} | "
            f"Principal Protected: {principal} | Surplus: 0"
        )

    if displacement_x < 0.50 or velocity_v > 0:
        return (
            f"[{tranche_id}] GREEN_LIGHT -> HOLD_CREST | "
            f"x={displacement_x:.2f}, v={velocity_v:.2f} | "
            f"Principal Protected: {principal} | Surplus: 0"
        )

    surplus = current - principal if current > principal else 0
    action = (
        f"ROLL_UNIT_SURPLUS -> {best_coupled_node}"
        if surplus > 0 and coupled_score >= 0.62
        else ("SWEEP_SOLANA_USDC" if surplus > 0 else "HOLD_PRINCIPAL")
    )

    return (
        f"[{tranche_id}] GREEN_LIGHT -> {action} | "
        f"FRI {composite_fri:.4f} >= {fri_threshold:.4f} | "
        f"Principal Protected: {principal} | Surplus: {surplus}"
    )


@tool
def record_finalized_tranche_outcome(
    tranche_id: str,
    source_node: str,
    target_node: str,
    composite_fri_at_entry: float,
    realized_edge_bps: float,
    principal_base_units: str,
    surplus_base_units: str,
    rolled_to_node: str | None = None,
) -> str:
    """
    Inserts a finalized unit-tranche outcome into Neon execution_feedback_loop
    using exact integer base units and updates neural_coupling_weights.
    """
    principal = int(principal_base_units)
    surplus = int(surplus_base_units)
    if principal < 0 or surplus < 0:
        raise ValueError("Unit amounts must be non-negative integers.")

    db_url = userdata.get("DATABASE_URL")
    if not db_url:
        raise ValueError("Missing DATABASE_URL in Colab Secrets.")

    is_win = realized_edge_bps > 0
    win_signal = 1.0 if is_win else 0.0
    alpha = 0.12
    weight_delta = max(-0.05, min(0.05, (realized_edge_bps / 1000.0) * alpha))
    zeta_delta = (-0.005 * alpha) if is_win else (0.015 * alpha)
    fri_adjust = -0.002 if is_win else 0.005

    conn = psycopg2.connect(db_url)
    try:
        with conn, conn.cursor() as cur:
            cur.execute(
                """
                INSERT INTO execution_feedback_loop (
                    unit_tranche_id, source_node, target_node,
                    composite_fri_at_entry, realized_edge_bps,
                    principal_protected, principal_units, surplus_units,
                    gate_status, rolled_to_node
                )
                VALUES (%s, %s, %s, %s, %s, TRUE, %s::NUMERIC(38, 0), %s::NUMERIC(38, 0), 'GREEN_LIGHT', %s)
                RETURNING outcome_id;
                """,
                (
                    tranche_id,
                    source_node,
                    target_node,
                    composite_fri_at_entry,
                    realized_edge_bps,
                    str(principal),
                    str(surplus),
                    rolled_to_node,
                ),
            )
            outcome_id = cur.fetchone()[0]

            cur.execute(
                """
                INSERT INTO neural_coupling_weights (
                    source_node, target_node, coupling_weight, damping_zeta,
                    fri_threshold, win_rate_ema, sample_count, last_calibrated_at
                )
                VALUES (%s, %s, GREATEST(0.05, 0.50 + %s), LEAST(0.95, GREATEST(0.05, 0.25 + %s)),
                        LEAST(0.55, GREATEST(0.30, 0.30 + %s)), %s, 1, NOW())
                ON CONFLICT (source_node, target_node) DO UPDATE SET
                    coupling_weight    = LEAST(1.50, GREATEST(0.05, neural_coupling_weights.coupling_weight + %s)),
                    damping_zeta       = LEAST(0.95, GREATEST(0.05, neural_coupling_weights.damping_zeta + %s)),
                    fri_threshold      = LEAST(0.55, GREATEST(0.30, neural_coupling_weights.fri_threshold + %s)),
                    win_rate_ema       = (neural_coupling_weights.win_rate_ema * (1 - %s)) + (%s * %s),
                    sample_count       = neural_coupling_weights.sample_count + 1,
                    last_calibrated_at = NOW();
                """,
                (
                    source_node,
                    target_node,
                    weight_delta,
                    zeta_delta,
                    fri_adjust,
                    win_signal,
                    weight_delta,
                    zeta_delta,
                    fri_adjust,
                    alpha,
                    win_signal,
                    alpha,
                ),
            )
    finally:
        conn.close()

    return (
        f"Recorded outcome_id={outcome_id} for {tranche_id} ({source_node}->{target_node}) | "
        f"edge_bps={realized_edge_bps} | principal={principal} | surplus={surplus}"
    )


# Quick local test of the handover tool (no DB needed)
print(
    evaluate_solana_unit_handover.invoke(
        {
            "tranche_id": "TRANCHE_UNIT_01",
            "composite_fri": 0.36,
            "fri_threshold": 0.30,
            "displacement_x": 0.58,
            "velocity_v": -0.04,
            "principal_base_units": "100000000",
            "current_base_units": "114500000",
        }
    )
)
````

#### 2. How the Row-Level Security (RLS) Policy Filters Data Dynamically

Row-Level Security allows us to apply security filters directly at the database engine level, guaranteeing that queries only return records the current user session is cleared to see.

##### Policy Architecture:
```sql
CREATE POLICY rls_clearance_policy ON intelligence_entities
FOR SELECT
USING (
    CASE current_setting('app.current_user_clearance', true)
        WHEN 'TOP_SECRET' THEN TRUE
        WHEN 'SECRET' THEN classification_level IN ('SECRET', 'CONFIDENTIAL')
        WHEN 'CONFIDENTIAL' THEN classification_level = 'CONFIDENTIAL'
        ELSE FALSE
    END
);
```

##### Dynamic Filtration Flow:
1. **User Authentication & Session Initialization:** Before retrieving any documents or vector matches, your application sets the session parameter using a transaction-local setting:
   `SET LOCAL app.current_user_clearance = 'SECRET';`
2. **Database-Level Isolation:**
   * If the session is authenticated as **`CONFIDENTIAL`**, the `CASE` statement only permits rows where `classification_level = 'CONFIDENTIAL'` to pass through. Any `SECRET` or `TOP_SECRET` rows are filtered out and remain completely invisible to the client.
   * If authenticated as **`SECRET`**, the session can read `CONFIDENTIAL` and `SECRET` rows.
   * Only an active session set to **`TOP_SECRET`** will bypass the filtering checks (`THEN TRUE`), exposing all records.
3. **Defense-in-Depth for Vectors:** Because the engine applies this filtering during table scanning, any high-dimensional vector similarity operations (`<=>` or `<->`) are only computed against records the session clearance authorizes. This prevents unauthorized side-channel leakages of classified summaries.

````python
import psycopg2
from google.colab import userdata

# Expanded SQL script for initializing the pgvector extension, creating intelligence schema, feedback loops, and RLS policies
initialize_schema_sql = """
-- 1. Enable pgvector extension
CREATE EXTENSION IF NOT EXISTS vector;

-- 2. Create classified intelligence entities table
CREATE TABLE IF NOT EXISTS intelligence_entities (
    entity_id UUID PRIMARY KEY DEFAULT gen_random_uuid(),
    entity_name VARCHAR(255) NOT NULL,
    entity_type VARCHAR(100) NOT NULL, -- e.g., 'PERSON', 'ORGANIZATION', 'FACILITY'
    classification_level VARCHAR(50) NOT NULL DEFAULT 'SECRET', -- e.g., 'CONFIDENTIAL', 'SECRET', 'TOP_SECRET'
    summary TEXT,
    entity_embedding vector(1536), -- Vector embeddings dimension
    created_at TIMESTAMP WITH TIME ZONE DEFAULT NOW()
);

-- 3. Create execution feedback loop table for Solana transactions
CREATE TABLE IF NOT EXISTS execution_feedback_loop (
    outcome_id SERIAL PRIMARY KEY,
    unit_tranche_id VARCHAR(100) NOT NULL,
    source_node VARCHAR(100) NOT NULL,
    target_node VARCHAR(100) NOT NULL,
    composite_fri_at_entry NUMERIC(10, 4) NOT NULL,
    realized_edge_bps NUMERIC(10, 4) NOT NULL,
    principal_protected BOOLEAN DEFAULT TRUE,
    principal_units NUMERIC(38, 0) NOT NULL,
    surplus_units NUMERIC(38, 0) NOT NULL,
    gate_status VARCHAR(50) DEFAULT 'GREEN_LIGHT',
    rolled_to_node VARCHAR(100),
    created_at TIMESTAMP WITH TIME ZONE DEFAULT NOW()
);

-- 4. Create neural coupling weights table
CREATE TABLE IF NOT EXISTS neural_coupling_weights (
    source_node VARCHAR(100) NOT NULL,
    target_node VARCHAR(100) NOT NULL,
    coupling_weight NUMERIC(10, 4) DEFAULT 0.50,
    damping_zeta NUMERIC(10, 4) DEFAULT 0.25,
    fri_threshold NUMERIC(10, 4) DEFAULT 0.35,
    win_rate_ema NUMERIC(10, 4) DEFAULT 0.00,
    sample_count INTEGER DEFAULT 0,
    last_calibrated_at TIMESTAMP WITH TIME ZONE DEFAULT NOW(),
    PRIMARY KEY (source_node, target_node)
);

-- 5. Enable Row-Level Security (RLS) strictly on intelligence entities
ALTER TABLE intelligence_entities ENABLE ROW LEVEL SECURITY;
ALTER TABLE intelligence_entities FORCE ROW LEVEL SECURITY;

-- 6. Create RLS policies to restrict row access dynamically based on custom settings
DROP POLICY IF EXISTS rls_clearance_policy ON intelligence_entities;
CREATE POLICY rls_clearance_policy ON intelligence_entities
FOR SELECT
USING (
    CASE current_setting('app.current_user_clearance', true)
        WHEN 'TOP_SECRET' THEN TRUE
        WHEN 'SECRET' THEN classification_level IN ('SECRET', 'CONFIDENTIAL')
        WHEN 'CONFIDENTIAL' THEN classification_level = 'CONFIDENTIAL'
        ELSE FALSE
    END
);
"""

print("Expanded Schema definition (Intelligence + Feedback loops + Neural weights) loaded.")
````

### Python snippet to initialize and query pgvector with psycopg2

````python
import numpy as np

def query_similar_entities(query_vector: list, user_clearance: str, limit: int = 5):
    db_url = userdata.get("DATABASE_URL")
    if not db_url:
        raise ValueError("Missing DATABASE_URL in Colab Secrets.")

    # Convert Python numeric list into standard PostgreSQL vector representation string
    vector_literal = "[" + ",".join(map(str, query_vector)) + "]"

    conn = psycopg2.connect(db_url)
    try:
        with conn, conn.cursor() as cur:
            # Configure user authorization level in this session
            cur.execute("SET LOCAL app.current_user_clearance = %s;", (user_clearance,))

            # Cosine similarity query using '<=>'
            cur.execute(
                """
                SELECT entity_id, entity_name, classification_level, entity_embedding <=> %s::vector AS distance
                FROM intelligence_entities
                ORDER BY distance ASC
                LIMIT %s;
                """,
                (vector_literal, limit)
            )

            results = cur.fetchall()
            return results
    finally:
        conn.close()

print("Query pipeline helper configured ready for execution.")
````

### 📐 System Architecture Summary

#### A. Solana Unit Handover Protocol
* **Crest-Gate Optimization:** Handover execution (`evaluate_solana_unit_handover`) protects principal assets under precise integer base units and ensures high-fidelity routing of accrued yield surplus.
* **Gate Triggers:** Execution relies on a composite FRI patience threshold. If conditions breach parameters (e.g., negative FRI thresholds), the system shifts state to `YELLOW_LIGHT_HOLD -> BLOCK_HOLD`.
* **Coupling Optimization:** When surplus exists, the algorithm rolls yield units dynamically via neural coupling paths (`ROLL_UNIT_SURPLUS`) when coupling scores satisfy target benchmarks ($\\ge 0.62$).

#### B. RLS & High-Dimensional Vector Architecture
* **Multi-Tenant Security Boundaries:** Utilizes native PostgreSQL Row-Level Security (RLS) policies acting at the transaction level (`SET LOCAL app.current_user_clearance`).
* **Vector Search Engine:** Leverages `pgvector` to run cosine similarity queries using the `<=>` distance operator. Due to RLS, queries filter and rank elements dynamically, preventing any data exposure outside designated classification constraints (`CONFIDENTIAL` $\\subset$ `SECRET` $\\subset$ `TOP_SECRET`).

````python
import psycopg2
from google.colab import userdata

def validate_rls_policies_live():
    """
    Connects to the live database, creates a temporary restricted non-owner role
    to properly test Row-Level Security (since table owners/superusers bypass RLS),
    and verifies data isolation boundaries with clean role disposal.
    """
    try:
        db_url = userdata.get("DATABASE_URL")
    except Exception as e:
        print("❌ DATABASE_URL is not set or accessible in Colab Secrets.")
        return

    if not db_url or db_url.strip() == "":
        print("❌ DATABASE_URL is empty in Colab Secrets.")
        return

    print("🔍 Validating Live Database RLS Rules (with Non-Owner Restricted Role via Session SET ROLE)...")
    try:
        conn = psycopg2.connect(db_url)
        conn.autocommit = True
        with conn.cursor() as cur:
            # 1. Check table existence
            cur.execute(
                """
                SELECT EXISTS (
                    SELECT FROM information_schema.tables
                    WHERE table_schema = 'public' AND table_name = 'intelligence_entities'
                );
                """
            )
            if not cur.fetchone()[0]:
                print("❌ Table 'intelligence_entities' does not exist! Please run schema setup first.")
                return

            # Get current session user to dynamically grant role membership
            cur.execute("SELECT CURRENT_USER;")
            current_session_user = cur.fetchone()[0]

            # 2. Setup a non-owner database role cleanly with explicit privileges
            try:
                cur.execute("RESET ROLE;")
                cur.execute(f"REVOKE rls_restricted_test_user FROM {current_session_user};")
            except Exception:
                pass
            try:
                cur.execute("REVOKE ALL ON ALL TABLES IN SCHEMA public FROM rls_restricted_test_user;")
                cur.execute("REVOKE ALL ON SCHEMA public FROM rls_restricted_test_user;")
                cur.execute("DROP ROLE IF EXISTS rls_restricted_test_user;")
            except Exception:
                pass

            cur.execute("CREATE ROLE rls_restricted_test_user WITH LOGIN PASSWORD 'test_rls_pass_9981';")
            cur.execute("GRANT USAGE ON SCHEMA public TO rls_restricted_test_user;")
            cur.execute("GRANT SELECT ON TABLE public.intelligence_entities TO rls_restricted_test_user;")

            # Grant the test role to our active connection user to allow 'SET ROLE'
            cur.execute(f"GRANT rls_restricted_test_user TO {current_session_user};")

            # 3. Use SET ROLE to transition current session to the restricted user
            cur.execute("SET ROLE rls_restricted_test_user;")
            cur.execute("SET search_path TO public;")

            try:
                # Test RLS for CONFIDENTIAL clearance
                cur.execute("SET LOCAL app.current_user_clearance = 'CONFIDENTIAL';")
                cur.execute("SELECT entity_name, classification_level FROM public.intelligence_entities;")
                conf_rows = cur.fetchall()
                print(f"✓ CONFIDENTIAL Clearance read {len(conf_rows)} records.")
                for name, level in conf_rows:
                    assert level == "CONFIDENTIAL", f"RLS Breach! Confidential clearance read a {level} row."

                # Test RLS for SECRET clearance
                cur.execute("SET app.current_user_clearance = 'SECRET';")
                cur.execute("SELECT entity_name, classification_level FROM public.intelligence_entities;")
                sec_rows = cur.fetchall()
                print(f"✓ SECRET Clearance read {len(sec_rows)} records.")
                for name, level in sec_rows:
                    assert level in ["CONFIDENTIAL", "SECRET"], f"RLS Breach! Secret clearance read a {level} row."

                # Test RLS for TOP_SECRET clearance
                cur.execute("SET app.current_user_clearance = 'TOP_SECRET';")
                cur.execute("SELECT entity_name, classification_level FROM public.intelligence_entities;")
                top_rows = cur.fetchall()
                print(f"✓ TOP_SECRET Clearance read {len(top_rows)} records.")

                print("\n✅ RLS Database Security Policies verified successfully against live database!")
            finally:
                # Reset role back to original connection owner for clean-up
                cur.execute("RESET ROLE;")

            # Clean up the test role dependencies and drop
            cur.execute(f"REVOKE rls_restricted_test_user FROM {current_session_user};")
            cur.execute("REVOKE SELECT ON TABLE public.intelligence_entities FROM rls_restricted_test_user;")
            cur.execute("REVOKE USAGE ON SCHEMA public FROM rls_restricted_test_user;")
            cur.execute("DROP ROLE IF EXISTS rls_restricted_test_user;")

    except Exception as e:
        print(f"❌ RLS Live validation failed: {e}")
    finally:
        if 'conn' in locals() and conn:
            conn.close()

# Run live RLS check
validate_rls_policies_live()
````

````python
import time
import matplotlib.pyplot as plt

def profile_clearance_performance_impact():
    """
    Simulates database query scanning over different clearance boundaries.
    We measure processing latency changes based on how much of the index space
    must be searched/filtered relative to the current session clearance.
    """
    clearances = ["CONFIDENTIAL", "SECRET", "TOP_SECRET"]
    latencies = []

    # Simulate multiple runs to capture average lookup overheads
    for cl in clearances:
        start_time = time.perf_counter()
        for _ in range(5000):
            # Simulate work (shorter paths are evaluated for more restrictive settings)
            if cl == "CONFIDENTIAL":
                _ = [x for x in range(100)]
            elif cl == "SECRET":
                _ = [x for x in range(300)]
            else: # TOP_SECRET accesses all records, wider search area
                _ = [x for x in range(600)]
        end_time = time.perf_counter()
        latencies.append((end_time - start_time) * 1000) # milliseconds

    # Plot performance results
    plt.figure(figsize=(8, 4))
    bars = plt.bar(clearances, latencies, color=['#4caf50', '#ff9800', '#f44336'], edgecolor='black', width=0.5)
    plt.title("Impact of Clearance Level Scope on Search Latency (Simulation)", fontsize=12, fontweight='bold')
    plt.ylabel("Cumulative Query Execution Time (ms)", fontsize=10)
    plt.xlabel("Session Clearance Tier", fontsize=10)
    plt.grid(axis='y', linestyle='--', alpha=0.5)

    # Annotate bar values
    for bar in bars:
        yval = bar.get_height()
        plt.text(bar.get_x() + bar.get_width()/2.0, yval + (max(latencies)*0.02), f"{yval:.2f} ms", ha='center', va='bottom', fontweight='bold')

    plt.tight_layout()
    plt.show()

profile_clearance_performance_impact()
````
